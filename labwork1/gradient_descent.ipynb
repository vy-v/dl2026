```python
def f(x):
    return x**2
```

```python
def df(x):
    return 2*x
```

```python
def gradient_descent(x0, lr, max_iter=1000, threshold=0.01):
    x = x0
    t = 1

    while t <= max_iter:
        fx = f(x)
        x_new = x - lr * df(x)
        fx_new = f(x_new)
        
        print(f"Time = {t} | x = {x:.4f} | f(x) = {fx:.4f}")

        if abs(fx_new - fx) < threshold:
            print(f"\n Converged after {t} iterations, f(x) = {fx_new:.4f}")
            break

        x = x_new
        t += 1
    else:
        print(f"\n Reached max iterations ({max_iter}), f(x) = {f(x):.4f}")

```

```python
x0 = 10.0
learning_rates = [1.0, 0.1, 0.01, 0.001]

for lr in learning_rates:
    print(f"  lr = {lr}")
    gradient_descent(x0=x0, lr=lr)
```
